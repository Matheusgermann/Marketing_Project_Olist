
# Infrastructure Setup
Creating the dedicated schemas (Bronze, Silver, and Gold) within the workspace catalog to ensure data isolation and governance.

```python
%sql
-- Creating Schemas for Medallion Architecture
CREATE SCHEMA IF NOT EXISTS workspace.bronze_marketing_project;
CREATE SCHEMA IF NOT EXISTS workspace.silver_marketing_project;
CREATE SCHEMA IF NOT EXISTS workspace.gold_marketing_project;
```


# Data Governance: Schemas
Documenting each schema for the analytics team.

```python
%sql
COMMENT ON SCHEMA workspace.bronze_marketing_project IS 'Landing Zone: Raw unchanged data directly from Kaggle Olist Dataset';
COMMENT ON SCHEMA workspace.silver_marketing_project IS 'Single Source of Truth: Cleaned, filtered, and deduplicated business data';
COMMENT ON SCHEMA workspace.gold_marketing_project IS 'Executive Metrics: Aggregated KPIs and RFM Segments for Power BI consumption';
```


# Volume Creation
Defining a secure Unity Catalog Volume to serve as the landing zone for raw files (CSVs) before ingestion.

```python
%sql
-- Creating a secure landing zone for raw files
CREATE VOLUME IF NOT EXISTS workspace.bronze_marketing_project.raw_data;
```
